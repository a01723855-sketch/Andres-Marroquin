```python
import pandas as pd

# Load the anthropometric dataset
anthro = pd.read_csv('/content/antropometria-dataset-2.csv')
anthro.head()
```

```python
# Check the columns and data types before doing the analysis
anthro.info()
```

## Mean Weight by State and Sex

I grouped the weight data by state and sex to compare the mean weight for each group. I also included the mean weight for all observations in each state.

```python
weight_by_group = anthro.pivot_table(
    index='desc_ent',
    columns='sexo',
    values='peso',
    aggfunc='mean'
)

weight_by_group.columns = ['Mean Weight (Sex 1)', 'Mean Weight (Sex 2)']

state_means = anthro.groupby('desc_ent')['peso'].mean().rename('Mean Weight (All)')

weight_summary = pd.concat([state_means, weight_by_group], axis=1)
weight_summary
```

## What the Table Shows

The resulting table makes it possible to compare average weight between states. It also separates the results by sex, so the differences between the two groups can be seen within the same state. The overall column gives a general average without separating the observations by sex.

These values can be useful for identifying differences in the dataset between locations and between the two sex categories. However, the table only shows averages, so it should not be used by itself to explain why those differences occur.

## Marital Status in the Adult Dataset

For the second part, I used the adult dataset to count how many records belong to each marital-status category. I also calculated each category as a percentage of the total.

```python
adult = pd.read_csv('/content/adult.csv')
adult.head()
```

```python
# Review the structure of the second dataset
adult.info()
```

```python
status_table = adult['marital-status'].value_counts().rename_axis('Marital Status').reset_index(name='Count')
status_table['Percent'] = status_table['Count'].div(status_table['Count'].sum()).mul(100)
status_table
```

## Specific State Check

```python
morelos_mean = weight_summary.loc['MORELOS', 'Mean Weight (All)']
print(f'Mean weight for MORELOS: {morelos_mean:.4f}')
```
